## 1. Mean

**Function:** `stat_mean(x)`

Core statistical calculation implemented directly in Python.

```python
def stat_mean(x):
    return sum(x) / len(x)
```

```python
mean_data = [2, 4, 4, 6, 9]
mean_value = stat_mean(mean_data)
print(mean_value)
```

```python
plt.bar(range(len(mean_data)), mean_data)
plt.axhline(mean_value, linestyle='--', label=f'Mean = {mean_value:.2f}')
plt.title('1. Mean'); plt.xlabel('Observation'); plt.ylabel('Value'); plt.legend(); plt.show()
```

## 2. Median

**Function:** `stat_median(x)`

Core statistical calculation implemented directly in Python.

```python
def stat_median(x):
    s = sorted(x)
        n = len(s)
        mid = n // 2
        return s[mid] if n % 2 else (s[mid-1] + s[mid]) / 2
```

```python
median_data = [1, 3, 4, 8, 10, 12]
median_value = stat_median(median_data)
print(median_value)
```

```python
plt.hist(median_data, bins=6, edgecolor='black')
plt.axvline(median_value, linestyle='--', label=f'Median = {median_value:.1f}')
plt.title('2. Median'); plt.legend(); plt.show()
```

## 3. Mode

**Function:** `stat_mode(x)`

Core statistical calculation implemented directly in Python.

```python
def stat_mode(x):
    counts = {v: x.count(v) for v in set(x)}
        highest = max(counts.values())
        return sorted(v for v, count in counts.items() if count == highest)
```

```python
mode_data = [1, 2, 2, 3, 3, 3, 4]
mode_value = stat_mode(mode_data)
print(mode_value)
```

```python
plt.hist(mode_data, bins=range(1, 7), align='left', rwidth=.8)
plt.title(f'3. Mode(s): {mode_value}'); plt.xlabel('Value'); plt.ylabel('Frequency'); plt.show()
```

## 4. Variance

**Function:** `stat_variance(x, sample=True)`

Core statistical calculation implemented directly in Python.

```python
def stat_variance(x, sample=True):
    mu = sum(x) / len(x)
        divisor = len(x) - 1 if sample else len(x)
        return sum((v - mu) ** 2 for v in x) / divisor
```

```python
variance_data = [4, 5, 7, 8, 11]
variance_value = stat_variance(variance_data)
print(variance_value)
```

```python
plt.bar(range(len(variance_data)), [(v - sum(variance_data)/len(variance_data))**2 for v in variance_data])
plt.title(f'4. Squared deviations; sample variance={variance_value:.2f}'); plt.xlabel('Observation'); plt.ylabel('Squared deviation'); plt.show()
```

## 5. Standard Deviation

**Function:** `stat_std(x, sample=True)`

Core statistical calculation implemented directly in Python.

```python
def stat_std(x, sample=True):
    return math.sqrt(stat_variance(x, sample=sample))
```

```python
std_data = [4, 5, 7, 8, 11]
std_value = stat_std(std_data)
print(std_value)
```

```python
mu = sum(std_data)/len(std_data)
plt.hist(std_data, bins=5, edgecolor='black')
plt.axvline(mu, label='Mean')
plt.axvline(mu+std_value, linestyle='--', label='+1 SD'); plt.axvline(mu-std_value, linestyle='--', label='-1 SD')
plt.title(f'5. Standard deviation = {std_value:.2f}'); plt.legend(); plt.show()
```

## 6. Probability

**Function:** `stat_probability(favorable, total)`

Core statistical calculation implemented directly in Python.

```python
def stat_probability(favorable, total):
    if total <= 0 or not 0 <= favorable <= total:
            raise ValueError('Require 0 <= favorable <= total and total > 0')
        return favorable / total
```

```python
p = stat_probability(3, 10)
print(f'Probability: {p:.0%}')
```

```python
plt.bar(['Favorable','Other'], [3,7]); plt.title(f'6. Probability = {p:.0%}'); plt.ylabel('Count'); plt.show()
```

## 7. Bayes' Theorem

**Function:** `stat_bayes(prior, sensitivity, false_positive_rate)`

Core statistical calculation implemented directly in Python.

```python
def stat_bayes(prior, sensitivity, false_positive_rate):
    denominator = sensitivity * prior + false_positive_rate * (1-prior)
        if denominator == 0:
            raise ValueError('Evidence has zero probability')
        return sensitivity * prior / denominator
```

```python
posterior = stat_bayes(.01, .95, .05)
print(f'Posterior probability: {posterior:.2%}')
```

```python
plt.bar(['Prior','Posterior'], [.01, posterior]); plt.ylim(0,1); plt.ylabel('Probability'); plt.title('7. Bayes theorem: positive-test example'); plt.show()
```

## 8. Correlation

**Function:** `stat_correlation(x, y)`

Core statistical calculation implemented directly in Python.

```python
def stat_correlation(x, y):
    if len(x) != len(y) or len(x) < 2:
            raise ValueError('Inputs must have equal length >= 2')
        mx, my = sum(x)/len(x), sum(y)/len(y)
        dx, dy = [v-mx for v in x], [v-my for v in y]
        denom = math.sqrt(sum(v*v for v in dx) * sum(v*v for v in dy))
        if denom == 0:
            raise ValueError('Correlation undefined for constant input')
        return sum(a*b for a,b in zip(dx,dy)) / denom
```

```python
corr_x=[1,2,3,4,5,6]; corr_y=[2,3,5,4,6,8]
r=stat_correlation(corr_x,corr_y)
print(f'Pearson r = {r:.3f}')
```

```python
plt.scatter(corr_x,corr_y); plt.title(f'8. Correlation r = {r:.3f}'); plt.xlabel('X'); plt.ylabel('Y'); plt.show()
```

## 9. Regression

**Function:** `stat_linear_regression(x, y)`

Core statistical calculation implemented directly in Python.

```python
def stat_linear_regression(x, y):
    if len(x) != len(y) or len(x) < 2:
            raise ValueError('Inputs must have equal length >= 2')
        mx, my = sum(x)/len(x), sum(y)/len(y)
        denom = sum((v-mx)**2 for v in x)
        if denom == 0:
            raise ValueError('X must vary')
        slope = sum((a-mx)*(b-my) for a,b in zip(x,y)) / denom
        intercept = my - slope*mx
        return slope, intercept
```

```python
reg_x=[1,2,3,4,5,6]; reg_y=[2,3,5,4,6,8]
slope, intercept=stat_linear_regression(reg_x,reg_y)
print(f'y = {slope:.2f}x + {intercept:.2f}')
```

```python
plt.scatter(reg_x,reg_y,label='Data'); plt.plot(reg_x,[slope*x+intercept for x in reg_x],label='Least-squares line'); plt.title('9. Linear regression'); plt.legend(); plt.show()
```

## 10. Normal Distribution

**Function:** `stat_normal_pdf(x, mu=0, sigma=1)`

Core statistical calculation implemented directly in Python.

```python
def stat_normal_pdf(x, mu=0, sigma=1):
    if sigma <= 0:
            raise ValueError('sigma must be positive')
        z = (x-mu)/sigma
        return math.exp(-0.5*z*z)/(sigma*math.sqrt(2*math.pi))
```

```python
normal_x=[i/10 for i in range(-40,41)]
normal_y=[stat_normal_pdf(x,0,1) for x in normal_x]
```

```python
plt.plot(normal_x,normal_y); plt.title('10. Standard normal probability density'); plt.xlabel('x'); plt.ylabel('Density'); plt.show()
```

## 11. Hypothesis Testing

**Function:** `stat_one_sample_t(sample, null_mean=0)`

Core statistical calculation implemented directly in Python.

```python
def stat_one_sample_t(sample, null_mean=0):
    if len(sample) < 2:
            raise ValueError('Need at least two observations')
        mean = sum(sample)/len(sample)
        sd = stat_std(sample, sample=True)
        se = sd/math.sqrt(len(sample))
        return (mean-null_mean)/se
```

```python
test_sample=[104,101,99,107,103,105,102,106]
t_stat=stat_one_sample_t(test_sample,100)
print(f't statistic = {t_stat:.3f}')
```

```python
plt.hist(test_sample,bins=6,edgecolor='black'); plt.axvline(100,linestyle='--',label='Null mean'); plt.axvline(sum(test_sample)/len(test_sample),label='Sample mean'); plt.title('11. Hypothesis test sample'); plt.legend(); plt.show()
```

## 12. P-value

**Function:** `stat_two_sided_normal_p(z)`

Core statistical calculation implemented directly in Python.

```python
def stat_two_sided_normal_p(z):
    return math.erfc(abs(z) / math.sqrt(2))  # Normal approximation; not an exact t-test p-value
```

```python
z=2.0
p_value=stat_two_sided_normal_p(z)
print(f'Approximate two-sided p-value: {p_value:.4f}')
```

```python
xs=[i/100 for i in range(-400,401)]
ys=[math.exp(-x*x/2)/math.sqrt(2*math.pi) for x in xs]
plt.plot(xs,ys); plt.axvline(z,linestyle='--'); plt.axvline(-z,linestyle='--'); plt.title(f'12. Normal-tail approximation; p ≈ {p_value:.4f}'); plt.show()
```

## 13. Confidence Interval

**Function:** `stat_mean_ci(sample, z=1.96)`

Core statistical calculation implemented directly in Python.

```python
def stat_mean_ci(sample, z=1.96):
    if len(sample) < 2:
            raise ValueError('Need at least two observations')
        mean = sum(sample)/len(sample)
        margin = z * stat_std(sample, sample=True)/math.sqrt(len(sample))
        return mean-margin, mean+margin
```

```python
ci_data=[12,14,13,15,11,16,14,13]
ci_low,ci_high=stat_mean_ci(ci_data)
print(f'Approximate 95% CI: [{ci_low:.2f}, {ci_high:.2f}]')
```

```python
plt.errorbar([sum(ci_data)/len(ci_data)],[0],xerr=[[(sum(ci_data)/len(ci_data))-ci_low],[ci_high-(sum(ci_data)/len(ci_data))]],fmt='o',capsize=8); plt.yticks([]); plt.title('13. Approximate 95% confidence interval'); plt.xlabel('Mean'); plt.show()
```

## 14. T-test

**Function:** `stat_welch_t(a,b)`

Core statistical calculation implemented directly in Python.

```python
def stat_welch_t(a,b):
    if len(a)<2 or len(b)<2:
            raise ValueError('Each group needs at least two values')
        ma, mb = sum(a)/len(a), sum(b)/len(b)
        va, vb = stat_variance(a), stat_variance(b)
        return (ma-mb)/math.sqrt(va/len(a)+vb/len(b))
```

```python
group_a=[78,82,85,88,90,84]; group_b=[72,75,79,80,77,74]
t_welch=stat_welch_t(group_a,group_b)
print(f'Welch t statistic = {t_welch:.3f}')
```

```python
plt.boxplot([group_a,group_b],tick_labels=['A','B']); plt.title('14. Welch two-sample t-test'); plt.ylabel('Measurement'); plt.show()
```

## 15. ANOVA

**Function:** `stat_one_way_anova(*groups)`

Core statistical calculation implemented directly in Python.

```python
def stat_one_way_anova(*groups):
    if len(groups)<2 or any(len(g)<2 for g in groups):
            raise ValueError('Need at least two groups, each with >=2 values')
        all_values=[v for g in groups for v in g]
        grand=sum(all_values)/len(all_values)
        ss_between=sum(len(g)*(sum(g)/len(g)-grand)**2 for g in groups)
        ss_within=sum(sum((v-sum(g)/len(g))**2 for v in g) for g in groups)
        df_between=len(groups)-1; df_within=len(all_values)-len(groups)
        return (ss_between/df_between)/(ss_within/df_within)
```

```python
anova_groups=[[10,11,9,10,12],[13,14,12,15,13],[16,15,17,14,16]]
f_stat=stat_one_way_anova(*anova_groups)
print(f'One-way ANOVA F statistic = {f_stat:.3f}')
```

```python
plt.boxplot(anova_groups,tick_labels=['Control','A','B']); plt.title(f'15. One-way ANOVA; F = {f_stat:.2f}'); plt.ylabel('Outcome'); plt.show()
```

## 16. Chi-Square Test

**Function:** `stat_chi_square(observed)`

Core statistical calculation implemented directly in Python.

```python
def stat_chi_square(observed):
    rows=len(observed); cols=len(observed[0])
        if rows<2 or cols<2 or any(len(row)!=cols for row in observed) or any(v<0 for row in observed for v in row):
            raise ValueError('Expected a rectangular nonnegative table >= 2x2')
        row_totals=[sum(row) for row in observed]
        col_totals=[sum(observed[i][j] for i in range(rows)) for j in range(cols)]
        total=sum(row_totals)
        expected=[[row_totals[i]*col_totals[j]/total for j in range(cols)] for i in range(rows)]
        return sum((observed[i][j]-expected[i][j])**2/expected[i][j] for i in range(rows) for j in range(cols))
```

```python
contingency=[[30,50],[70,50]]
chi=stat_chi_square(contingency)
print(f'Chi-square statistic = {chi:.3f}')
```

```python
plt.imshow(contingency); plt.colorbar(label='Count'); plt.xticks([0,1],['Control','New']); plt.yticks([0,1],['Clicked','Not clicked']); plt.title(f'16. Observed counts; χ²={chi:.2f}'); plt.show()
```

## 17. Decision Trees

**Function:** `stat_decision_stump(X,y)`

Small educational ML implementation, written for clarity rather than production scale.

```python
def stat_decision_stump(X,y):
    if not X or len(X)!=len(y):
            raise ValueError('X and y must be nonempty and aligned')
        best=None
        for feature in range(len(X[0])):
            values=sorted(set(row[feature] for row in X))
            for left,right in zip(values,values[1:]):
                threshold=(left+right)/2
                groups=[[y[i] for i,row in enumerate(X) if row[feature]<=threshold],[y[i] for i,row in enumerate(X) if row[feature]>threshold]]
                if not all(groups): continue
                impurity=sum(len(g)/len(y)*(1-sum(g.count(c)/len(g)**2 for c in set(g))) for g in groups)
                if best is None or impurity<best[0]: best=(impurity,feature,threshold)
        return best
```

```python
tree_X=[[1,2],[2,1],[3,3],[4,2],[5,5],[6,4]]; tree_y=[0,0,0,1,1,1]
stump=stat_decision_stump(tree_X,tree_y)
print('Best Gini split (impurity, feature, threshold):',stump)
```

```python
plt.scatter([r[0] for r in tree_X],[r[1] for r in tree_X],c=tree_y); plt.axvline(stump[2] if stump[1]==0 else 3.5,linestyle='--'); plt.title('17. Decision-stump training data'); plt.xlabel('Feature 1'); plt.ylabel('Feature 2'); plt.show()
```

## 18. K-Means Clustering

**Function:** `stat_kmeans(points,k,steps=30)`

Small educational ML implementation, written for clarity rather than production scale.

```python
def stat_kmeans(points,k,steps=30):
    if not points or not 1<=k<=len(points):
            raise ValueError('k must be between 1 and number of points')
        centers=[list(points[i]) for i in range(k)]
        for _ in range(steps):
            labels=[min(range(k),key=lambda j:sum((a-b)**2 for a,b in zip(p,centers[j]))) for p in points]
            updated=[]
            for j in range(k):
                cluster=[p for p,label in zip(points,labels) if label==j]
                updated.append([sum(p[d] for p in cluster)/len(cluster) if cluster else centers[j][d] for d in range(len(centers[j]))])
            if updated==centers: break
            centers=updated
        return labels,centers
```

```python
cluster_points=[[0,0],[.2,.1],[.1,.3],[5,5],[5.2,4.9],[4.8,5.1],[0,5],[.2,5.1],[.1,4.8]]
cluster_labels,cluster_centers=stat_kmeans(cluster_points,3)
print('Labels:',cluster_labels,'\nCenters:',cluster_centers)
```

```python
plt.scatter([p[0] for p in cluster_points],[p[1] for p in cluster_points],c=cluster_labels); plt.scatter([p[0] for p in cluster_centers],[p[1] for p in cluster_centers],marker='X',s=180); plt.title('18. K-means from scratch'); plt.show()
```

## 19. Principal Component Analysis (PCA)

**Function:** `stat_pca_2d(points, iterations=100)`

Small educational ML implementation, written for clarity rather than production scale.

```python
def stat_pca_2d(points, iterations=100):
    if not points or len(points[0])!=2:
            raise ValueError('This teaching implementation expects 2D points')
        mx=sum(p[0] for p in points)/len(points); my=sum(p[1] for p in points)/len(points)
        centered=[(p[0]-mx,p[1]-my) for p in points]
        a=sum(x*x for x,y in centered); b=sum(x*y for x,y in centered); d=sum(y*y for x,y in centered)
        vx,vy=1.0,1.0
        for _ in range(iterations):
            nx,ny=a*vx+b*vy,b*vx+d*vy
            norm=math.hypot(nx,ny)
            if norm==0: raise ValueError('Zero-variance data')
            vx,vy=nx/norm,ny/norm
        scores=[x*vx+y*vy for x,y in centered]
        return scores,(vx,vy)
```

```python
pca_points=[(1,2),(2,3),(3,5),(4,4),(5,7),(6,8)]
pca_scores,pca_axis=stat_pca_2d(pca_points)
print('First principal axis:',pca_axis)
```

```python
plt.scatter([p[0] for p in pca_points],[p[1] for p in pca_points]); plt.title('19. PCA input points'); plt.xlabel('Feature 1'); plt.ylabel('Feature 2'); plt.show()
```

## 20. Time Series Forecasting

**Function:** `stat_forecast(values,horizon=5)`

Small educational ML implementation, written for clarity rather than production scale.

```python
def stat_forecast(values,horizon=5):
    if len(values)<2 or horizon<1:
            raise ValueError('Need >=2 values and positive horizon')
        x=list(range(len(values))); mx=sum(x)/len(x); my=sum(values)/len(values)
        slope=sum((a-mx)*(b-my) for a,b in zip(x,values))/sum((a-mx)**2 for a in x)
        intercept=my-slope*mx
        return [intercept+slope*t for t in range(len(values),len(values)+horizon)]
```

```python
series=[100,103,105,108,110,114,115,119]
forecast=stat_forecast(series,4)
print('Forecast:',forecast)
```

```python
plt.plot(range(len(series)),series,marker='o',label='Observed'); plt.plot(range(len(series),len(series)+len(forecast)),forecast,marker='o',linestyle='--',label='Forecast'); plt.title('20. Linear-trend time-series forecast'); plt.xlabel('Period'); plt.ylabel('Value'); plt.legend(); plt.show()
```
