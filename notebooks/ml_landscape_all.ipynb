# Supervised Learning

Learn from labeled examples where the desired target is known.

**Example:** Predict customer churn from historical customers labeled churn/stay.


## Classification


### Decision Tree

Use interpretable feature splits to classify an outcome.


```python
from sklearn.tree import DecisionTreeClassifier
X=[[25,1],[45,0],[35,1],[50,0]]; y=[0,1,0,1]
m=DecisionTreeClassifier(max_depth=2,random_state=42).fit(X,y)
print(m.predict([[40,0]]))
```

### Logistic Regression

Estimate the probability of a binary outcome such as fraud.


```python
from sklearn.linear_model import LogisticRegression
X=[[.1],[.2],[.8],[.9]]; y=[0,0,1,1]
m=LogisticRegression().fit(X,y)
print(m.predict_proba([[.7]]))
```

### Support Vector Machine

Separate classes with a maximum-margin decision boundary.


```python
from sklearn.svm import SVC
X=[[1,1],[1,2],[4,4],[4,5]]; y=[0,0,1,1]
m=SVC(kernel='linear').fit(X,y)
print(m.predict([[3.5,4]]))
```

### Naive Bayes

Classify using simple probabilistic feature assumptions.


```python
from sklearn.naive_bayes import GaussianNB
X=[[1,20],[2,18],[8,3],[9,2]]; y=[0,0,1,1]
m=GaussianNB().fit(X,y)
print(m.predict([[7,4]]))
```

### Quadratic Discriminant Analysis (QDA)

Classify with a separate covariance matrix for each class, allowing quadratic decision boundaries.


```python
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
X=[[1,1],[1.5,2],[2,1],[5,5],[5.5,6],[6,5]]; y=[0,0,0,1,1,1]
m=QuadraticDiscriminantAnalysis().fit(X,y)
print(m.predict([[4.5,4.8]])); print(m.predict_proba([[4.5,4.8]]))
```

### k-Nearest Neighbors

Classify a sample from the labels of nearby examples.


```python
from sklearn.neighbors import KNeighborsClassifier
X=[[1,1],[1,2],[5,5],[6,5]]; y=[0,0,1,1]
m=KNeighborsClassifier(3).fit(X,y)
print(m.predict([[5,4]]))
```

## Regression


### Linear Regression

Predict a continuous value with a linear relationship.


```python
from sklearn.linear_model import LinearRegression
X=[[1000],[1500],[2000],[2500]]; y=[200000,280000,360000,440000]
m=LinearRegression().fit(X,y)
print(m.predict([[1800]]))
```

### Polynomial Regression

Model a curved continuous relationship.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
X=[[1],[2],[3],[4]]; y=[1,4,9,16]
p=PolynomialFeatures(2); m=LinearRegression().fit(p.fit_transform(X),y)
print(m.predict(p.transform([[5]])))
```

### Ridge Regression

Shrink coefficients with L2 regularization to reduce overfitting.


```python
from sklearn.linear_model import Ridge
X=[[1,1],[2,2],[3,4],[4,5]]; y=[2,4,7,9]
m=Ridge(alpha=1).fit(X,y)
print(m.predict([[5,6]]))
```

### Lasso Regression

Use L1 regularization that can drive weak coefficients to zero.


```python
from sklearn.linear_model import Lasso
X=[[1,0],[2,0],[3,1],[4,1]]; y=[2,4,7,9]
m=Lasso(alpha=.1).fit(X,y)
print(m.coef_,m.predict([[5,1]]))
```

# Unsupervised Learning

Discover useful structure in data that has no target labels.

**Example:** Group customers into natural behavior segments without predefined segment labels.


## Clustering


### K-Means

Partition observations into centroid-based clusters.


```python
from sklearn.cluster import KMeans
X=[[1,1],[1,2],[8,8],[9,8]]
m=KMeans(n_clusters=2,random_state=42,n_init='auto').fit(X)
print(m.labels_,m.cluster_centers_)
```

### Hierarchical Clustering

Build a nested hierarchy of clusters; inspect the tree or cut it at a chosen number of groups.


```python
from sklearn.cluster import AgglomerativeClustering
X=[[1,1],[1.2,1.1],[5,5],[5.2,5.1],[9,1]]
m=AgglomerativeClustering(n_clusters=3,linkage='ward').fit(X)
print(m.labels_)
```

### DBSCAN

Find dense irregular clusters and flag isolated observations as noise.


```python
from sklearn.cluster import DBSCAN
X=[[1,1],[1.1,1.1],[8,8],[8.1,8.1],[20,20]]
print(DBSCAN(eps=.5,min_samples=2).fit(X).labels_)
```

### Mean Shift

Discover cluster centers by moving candidates toward dense regions.


```python
from sklearn.cluster import MeanShift
X=[[1,1],[1.2,1.1],[8,8],[8.2,8.1]]
m=MeanShift().fit(X)
print(m.labels_,m.cluster_centers_)
```

### Fuzzy C-Means

Give each sample partial membership in multiple clusters.


```python
import numpy as np
membership=np.array([[.9,.1],[.6,.4],[.2,.8]])
print(membership)
print('hard labels:',membership.argmax(1))
```

## Dimensionality Reduction


### PCA

Compress correlated numeric features into variance-preserving components.


```python
from sklearn.decomposition import PCA
X=[[1,2,3],[2,3,4],[8,9,10],[9,10,11]]
p=PCA(n_components=2).fit(X)
print(p.transform(X)); print(p.explained_variance_ratio_)
```

### t-SNE

Project high-dimensional samples into 2D for local-neighborhood visualization.


```python
from sklearn.manifold import TSNE
from sklearn.datasets import load_iris
X=load_iris().data[:40]
print(TSNE(n_components=2,perplexity=5,random_state=42).fit_transform(X)[:5])
```

### LDA

Project labeled samples into directions that best separate classes.


```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.datasets import load_iris
X,y=load_iris(return_X_y=True)
print(LinearDiscriminantAnalysis(n_components=2).fit_transform(X,y)[:5])
```

## Pattern Search


### Apriori

Find item combinations that frequently occur together.


```python
from itertools import combinations
tx=[{'milk','bread'},{'milk','eggs'},{'milk','bread','eggs'}]
items=sorted(set().union(*tx))
for p in combinations(items,2): print(p,sum(set(p)<=t for t in tx)/len(tx))
```

### FP-Growth

Mine frequent itemsets efficiently with a compressed prefix structure.


```python
tx=[['a','b'],['a','c'],['a','b','c']]
counts={}
for row in tx:
    for item in row: counts[item]=counts.get(item,0)+1
print(counts)
```

### ECLAT

Mine frequent itemsets by intersecting transaction-ID sets.


```python
tid={'milk':{1,2,3},'bread':{1,3},'eggs':{2,3}}
pair=tid['milk'] & tid['bread']
print(len(pair),pair)
```

# Semi-Supervised Learning

Use a small labeled set together with a much larger unlabeled set.

**Example:** Train an image classifier using 500 labeled images plus 50,000 unlabeled images.


## Pseudo-Labeling


### Self-Training

Use confident model predictions as pseudo-labels for unlabeled samples.


```python
from sklearn.semi_supervised import SelfTrainingClassifier
from sklearn.svm import SVC
X=[[1],[2],[3],[8],[9],[10]]; y=[0,0,-1,-1,1,1]
m=SelfTrainingClassifier(SVC(probability=True,gamma='auto')).fit(X,y)
print(m.transduction_)
```

# Self-Supervised Learning

Create supervision from the raw data itself to learn useful representations.

**Example:** Mask part of a sentence or image and train the model to recover what is missing.


## Representation Learning


### Masked Prediction

Hide part of an input and predict it from context.


```python
import numpy as np
x=np.array([10.,20.,30.,40.]); idx=2
context=np.delete(x,idx)
print('target:',x[idx],'toy prediction:',context.mean())
```

### Contrastive Learning

Pull related representations together and push unrelated ones apart.


```python
import numpy as np
a=np.array([1.,0.]); pos=np.array([.9,.1]); neg=np.array([-1.,0.])
cos=lambda x,y:x@y/(np.linalg.norm(x)*np.linalg.norm(y))
print(cos(a,pos),cos(a,neg))
```

# Reinforcement Learning

Learn behavior by interacting with an environment and maximizing cumulative reward.

**Example:** Train an agent to choose actions that maximize long-term reward in a game or control system.


## Value-Based


### Q-Learning

Learn a table of state-action values.


```python
import numpy as np
Q=np.zeros((3,2)); s,a,r,s2=0,1,1.,2; alpha,gamma=.1,.9
Q[s,a]+=alpha*(r+gamma*Q[s2].max()-Q[s,a])
print(Q)
```

### Deep Q-Network

Approximate Q-values with a neural network.


```python
import torch, torch.nn as nn
q=nn.Sequential(nn.Linear(4,16),nn.ReLU(),nn.Linear(16,2))
print(q(torch.tensor([[.1,.2,.3,.4]])).detach().numpy())
```

## Actor-Critic


### A3C

Use parallel actor-learners to improve a shared policy and value function.


```python
import torch.nn as nn
class ActorCritic(nn.Module):
    def __init__(self):
        super().__init__(); self.body=nn.Sequential(nn.Linear(4,32),nn.ReLU()); self.policy=nn.Linear(32,2); self.value=nn.Linear(32,1)
print(ActorCritic())
```

## Evolutionary


### Genetic Algorithm

Search for high-fitness solutions through selection and mutation.


```python
import random
pop=[random.uniform(-5,5) for _ in range(8)]; fitness=lambda x:-(x-2.)**2
for _ in range(5):
    pop=sorted(pop,key=fitness,reverse=True)[:4]; pop += [x+random.gauss(0,.5) for x in pop]
print(max(pop,key=fitness))
```

### Deep Belief Network (DBN)

A DBN stacks generative RBM layers. This example performs greedy layer-wise feature learning with two BernoulliRBM layers on a synthetic binary dataset.

**Try changing:** the hidden width, number of units, or batch size, then rerun and inspect the output.

```python
import numpy as np
from sklearn.neural_network import BernoulliRBM
rng = np.random.default_rng(12)
X = rng.binomial(1, 0.5, size=(256, 20)).astype(np.float32)
rbm1 = BernoulliRBM(n_components=16, learning_rate=0.05, n_iter=5, random_state=12)
rbm2 = BernoulliRBM(n_components=8, learning_rate=0.05, n_iter=5, random_state=13)
h1 = rbm1.fit_transform(X)
h2 = rbm2.fit_transform(h1)
assert h1.shape == (256, 16) and h2.shape == (256, 8)
print("Input / hidden layer shapes:", X.shape, h1.shape, h2.shape)
```

### Restricted Boltzmann Machine (RBM)

An RBM is an energy-based neural model with visible and hidden units and no within-layer connections. Fit the model and inspect the hidden-unit probabilities it learns.

**Try changing:** the hidden width, number of units, or batch size, then rerun and inspect the output.

```python
import numpy as np
from sklearn.neural_network import BernoulliRBM
rng = np.random.default_rng(21)
X = rng.binomial(1, 0.5, size=(300, 24)).astype(np.float32)
rbm = BernoulliRBM(n_components=10, learning_rate=0.05, n_iter=10, random_state=21)
rbm.fit(X)
hidden = rbm.transform(X)
assert hidden.shape == (300, 10)
print("Visible:", X.shape, "Hidden:", hidden.shape, "Mean activation:", round(float(hidden.mean()), 3))
```

### Inception Network (GoogLeNet-style)

Inception processes an input through parallel convolution and pooling branches, then concatenates their feature maps. This compact PyTorch block demonstrates the multi-scale pattern.

**Try changing:** the hidden width, number of units, or batch size, then rerun and inspect the output.

```python
import torch
from torch import nn
class InceptionBlock(nn.Module):
    def __init__(self, in_channels, width=8):
        super().__init__()
        self.branch1 = nn.Conv2d(in_channels, width, 1)
        self.branch3 = nn.Sequential(nn.Conv2d(in_channels, width, 1), nn.ReLU(), nn.Conv2d(width, width, 3, padding=1))
        self.branch_pool = nn.Sequential(nn.AvgPool2d(3, stride=1, padding=1), nn.Conv2d(in_channels, width, 1))
    def forward(self, x):
        return torch.relu(torch.cat([self.branch1(x), self.branch3(x), self.branch_pool(x)], dim=1))
x = torch.randn(4, 16, 32, 32)
y = InceptionBlock(16)(x)
assert y.shape == (4, 24, 32, 32)
print("Input:", tuple(x.shape), "Output:", tuple(y.shape))
```

### Deep Q-Network (DQN)

DQN uses a neural network to estimate a Q-value for each available action. This example builds and tests the value network; a full training agent also uses replay memory, a target network, and exploration.

**Try changing:** the hidden width, number of units, or batch size, then rerun and inspect the output.

```python
import torch
from torch import nn
class QNetwork(nn.Module):
    def __init__(self, state_dim, action_count):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(state_dim, 64), nn.ReLU(), nn.Linear(64, 64), nn.ReLU(), nn.Linear(64, action_count))
    def forward(self, state):
        return self.net(state)
q_net = QNetwork(state_dim=8, action_count=4)
states = torch.randn(32, 8)
q_values = q_net(states)
actions = q_values.argmax(dim=1)
assert q_values.shape == (32, 4) and actions.shape == (32,)
print("Q-value shape:", tuple(q_values.shape), "Action shape:", tuple(actions.shape))
```

### Generative Adversarial Network (GAN)

A GAN trains a generator to create samples and a discriminator to distinguish generated samples from real data. This small PyTorch example executes one update for each network.

**Try changing:** the hidden width, number of units, or batch size, then rerun and inspect the output.

```python
import torch
from torch import nn
z_dim, data_dim, batch = 16, 8, 32
generator = nn.Sequential(nn.Linear(z_dim, 32), nn.ReLU(), nn.Linear(32, data_dim))
discriminator = nn.Sequential(nn.Linear(data_dim, 32), nn.LeakyReLU(0.2), nn.Linear(32, 1), nn.Sigmoid())
opt_g = torch.optim.Adam(generator.parameters(), lr=1e-3)
opt_d = torch.optim.Adam(discriminator.parameters(), lr=1e-3)
loss_fn = nn.BCELoss()
real = torch.randn(batch, data_dim) * 0.5 + 1.0
ones, zeros = torch.ones(batch, 1), torch.zeros(batch, 1)
fake = generator(torch.randn(batch, z_dim)).detach()
loss_d = loss_fn(discriminator(real), ones) + loss_fn(discriminator(fake), zeros)
opt_d.zero_grad(); loss_d.backward(); opt_d.step()
fake = generator(torch.randn(batch, z_dim))
loss_g = loss_fn(discriminator(fake), ones)
opt_g.zero_grad(); loss_g.backward(); opt_g.step()
assert torch.isfinite(loss_d) and torch.isfinite(loss_g)
print(f"Discriminator loss={loss_d.item():.4f}; Generator loss={loss_g.item():.4f}")
```

# Ensemble Learning

Combine multiple learners to improve accuracy, stability, or generalization.

**Example:** Combine several models so the ensemble is more reliable than any single learner.


## Bagging


### Random Forest

Average many randomized trees to reduce variance.


```python
from sklearn.ensemble import RandomForestClassifier
X=[[1,1],[1,2],[4,4],[5,5]]; y=[0,0,1,1]
m=RandomForestClassifier(n_estimators=50,random_state=42).fit(X,y)
print(m.predict([[4,5]]))
```

## Boosting


### Gradient Boosting

Add learners sequentially to correct previous residual errors.


```python
from sklearn.ensemble import GradientBoostingClassifier
X=[[1],[2],[3],[8],[9],[10]]; y=[0,0,0,1,1,1]
m=GradientBoostingClassifier(random_state=42).fit(X,y)
print(m.predict([[7]]))
```

### XGBoost

Use regularized optimized gradient-boosted trees.


```python
from xgboost import XGBClassifier
X=[[1],[2],[3],[8],[9],[10]]; y=[0,0,0,1,1,1]
m=XGBClassifier(n_estimators=20,max_depth=2,eval_metric='logloss').fit(X,y)
print(m.predict([[7]]))
```

### LightGBM

Use efficient histogram-based gradient-boosted trees.


```python
from lightgbm import LGBMClassifier
X=[[1],[2],[3],[8],[9],[10]]; y=[0,0,0,1,1,1]
m=LGBMClassifier(n_estimators=20,verbosity=-1).fit(X,y)
print(m.predict([[7]]))
```

## Stacking


### Stacking Classifier

Train a meta-model on predictions from diverse base models.


```python
from sklearn.ensemble import StackingClassifier,RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
X=[[1],[2],[3],[8],[9],[10]]; y=[0,0,0,1,1,1]
m=StackingClassifier([('rf',RandomForestClassifier(n_estimators=20,random_state=42)),('svm',SVC(probability=True))],final_estimator=LogisticRegression()).fit(X,y)
print(m.predict([[7]]))
```

# Statistical Inference

Estimate uncertainty and draw population-level conclusions from sample data.

**Example:** Test whether an observed treatment effect is likely larger than ordinary sampling noise.


## Estimation


### Confidence Interval

Estimate a plausible range for an unknown population parameter.


```python
import numpy as np
from scipy import stats
x=np.array([10,11,9,12,10,11,13,10]); mean=x.mean(); margin=stats.t.ppf(.975,len(x)-1)*stats.sem(x)
print(mean-margin,mean+margin)
```

## Hypothesis Testing


### t-Test

Compare two means and quantify evidence against a no-difference hypothesis.


```python
from scipy.stats import ttest_ind
a=[10,11,9,10,12]; b=[14,13,15,14,16]
print(ttest_ind(a,b,equal_var=False))
```

## Bayesian Inference


### Bayes Update

Update a prior probability using observed evidence.


```python
prior=.01; sensitivity=.95; false_positive=.05
posterior=sensitivity*prior/(sensitivity*prior+false_positive*(1-prior))
print(round(posterior,4))
```
