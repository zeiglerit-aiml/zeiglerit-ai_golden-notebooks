## Gradient Descent

Run the cell below and experiment by changing values.

```python
w=0
lr=0.1
x=[1,2,3]
y=[2,4,6]
for _ in range(10):
    grad=sum((w*i-j)*i for i,j in zip(x,y))/len(x)
    w-=lr*grad
print(round(w,3))
```
