## Executable Python demonstrations (edit these functions to experiment)

```python
"""One runnable minimal learning experiment per algorithm, with a visualizable result."""
import numpy as np
from sklearn.datasets import make_regression, make_classification, make_blobs, make_moons
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, mean_squared_error, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor,RandomForestClassifier,HistGradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.neural_network import MLPClassifier
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
from sklearn.decomposition import PCA
import torch
from torch import nn

def run_algorithm(name, seed=42):
    """Returns (dataframe, metric string, chart title) for the selected algorithm."""
    import pandas as pd
    np.random.seed(seed); torch.manual_seed(seed)
    reg=name in ('Linear Regression','Random Forest Regressor')
    if reg:
        X,y=make_regression(n_samples=240,n_features=3,noise=19,random_state=seed)
        model=LinearRegression() if name=='Linear Regression' else RandomForestRegressor(n_estimators=60,random_state=seed)
        a,b=train_test_split(np.arange(len(y)),test_size=.25,random_state=seed)
        model.fit(X[a],y[a]); pred=model.predict(X[b]); score=f'Test RMSE: {np.sqrt(mean_squared_error(y[b],pred)):.3f}'
        return pd.DataFrame({'x':y[b], 'y':pred,'split':'test'}),score,'Actual vs predicted (ideal diagonal)'
    if name in ('K-Means','Hierarchical Clustering','DBSCAN','Principal Component Analysis','Autoencoder'):
        if name=='DBSCAN': X,_=make_moons(n_samples=260,noise=.07,random_state=seed)
        else: X,_=make_blobs(n_samples=260,centers=3,cluster_std=1.2,random_state=seed)
        X=StandardScaler().fit_transform(X)
        if name=='K-Means': labels=KMeans(n_clusters=3,n_init=10,random_state=seed).fit_predict(X)
        elif name=='Hierarchical Clustering': labels=AgglomerativeClustering(n_clusters=3).fit_predict(X)
        elif name=='DBSCAN': labels=DBSCAN(eps=.24,min_samples=5).fit_predict(X)
        elif name=='Principal Component Analysis':
            rng=np.random.default_rng(seed); X=np.column_stack([X,X[:,0]*.9+.1*rng.normal(size=len(X))]); z=PCA(n_components=2).fit_transform(X)
            return pd.DataFrame({'x':z[:,0],'y':z[:,1],'group':0}), 'Reduced 3 numeric features to 2 PCs','PCA projection'
        else:
            inp=torch.tensor(X,dtype=torch.float32); net=nn.Sequential(nn.Linear(2,4),nn.Tanh(),nn.Linear(4,1),nn.Tanh(),nn.Linear(1,4),nn.Tanh(),nn.Linear(4,2))
            opt=torch.optim.Adam(net.parameters(),lr=.02)
            for _ in range(100):
                opt.zero_grad(); pred=net(inp); loss=((pred-inp)**2).mean(); loss.backward();opt.step()
            with torch.no_grad(): recon=net(inp).numpy()
            err=np.sum((recon-X)**2,axis=1)
            return pd.DataFrame({'x':X[:,0],'y':X[:,1],'group':err}),f'Average squared reconstruction error: {np.mean(err):.4f}','Reconstruction error (continuous color)'
        groups=len(set(labels)-{-1}); metric=f'{groups} clusters found; noise label -1 when applicable'
        return pd.DataFrame({'x':X[:,0],'y':X[:,1],'group':labels}),metric,'Clusters in feature space'
    if name in ('Convolutional Neural Network','Recurrent Neural Network','Transformer Encoder'):
        n=160
        if name=='Convolutional Neural Network':
            X=torch.randn(n,1,8,8); y=(X[:,:,2:6,2:6].mean(dim=(1,2,3))>0).long()
            net=nn.Sequential(nn.Conv2d(1,4,3,padding=1),nn.ReLU(),nn.Flatten(),nn.Linear(4*8*8,2))
        else:
            X=torch.randn(n,8,3); y=(X[:,4:,0].mean(dim=1)>0).long()
            if name=='Recurrent Neural Network':
                class RNNModel(nn.Module):
                    def __init__(self): super().__init__();self.r=nn.RNN(3,12,batch_first=True);self.head=nn.Linear(12,2)
                    def forward(self,x): out,_=self.r(x);return self.head(out[:,-1,:])
                net=RNNModel()
            else:
                class TransformerModel(nn.Module):
                    def __init__(self):
                        super().__init__();self.proj=nn.Linear(3,12);self.pos=nn.Parameter(torch.zeros(1,8,12));self.attn=nn.TransformerEncoderLayer(d_model=12,nhead=3,batch_first=True,dropout=0.);self.head=nn.Linear(12,2)
                    def forward(self,x):return self.head(self.attn(self.proj(x)+self.pos)[:,4:,:].mean(dim=1))
                net=TransformerModel()
        train=torch.arange(120); test=torch.arange(120,160); opt=torch.optim.Adam(net.parameters(),lr=.02)
        for _ in range(45):
            net.train();opt.zero_grad();out=net(X[train]);loss=nn.functional.cross_entropy(out,y[train]);loss.backward();opt.step()
        net.eval()
        with torch.no_grad(): logits=net(X[test]); probs=logits.softmax(dim=1)[:,1].numpy()
        acc=accuracy_score(y[test].numpy(),probs>=.5)
        return pd.DataFrame({'x':np.arange(len(probs)),'y':probs,'group':y[test].numpy()}),f'Holdout accuracy: {acc:.2%} (toy dataset)','Predicted class-1 probabilities by sample'
    X,y=make_classification(n_samples=260,n_features=8,n_informative=5,n_redundant=1,random_state=seed)
    X=StandardScaler().fit_transform(X)
    if name=='Logistic Regression': model=LogisticRegression(max_iter=500)
    elif name=='Random Forest Classifier': model=RandomForestClassifier(n_estimators=60,random_state=seed)
    elif name=='Gradient Boosting': model=HistGradientBoostingClassifier(max_iter=75,random_state=seed)
    elif name=='Support Vector Machine': model=SVC(probability=True,random_state=seed)
    elif name=='K-Nearest Neighbors': model=KNeighborsClassifier(n_neighbors=7)
    elif name=='Naive Bayes': model=MultinomialNB();X=np.abs(X)
    elif name=='Neural Network (MLP)': model=MLPClassifier(hidden_layer_sizes=(16,),max_iter=250,random_state=seed)
    else: raise ValueError(f'Unknown algorithm: {name}')
    a,b=train_test_split(np.arange(len(y)),test_size=.25,random_state=seed,stratify=y)
    model.fit(X[a],y[a]); pred=model.predict(X[b]); probs=model.predict_proba(X[b])[:,1]
    return pd.DataFrame({'x':np.arange(len(b)),'y':probs,'group':y[b]}),f'Holdout accuracy: {accuracy_score(y[b],pred):.2%}','Predicted class-1 probabilities by sample'
```

## 1. Proximal Policy Optimization (PPO)

PPO is an on-policy policy-gradient method. The clipped objective limits how far a new policy can move from the behavior policy in one update. This compact example computes the clipped surrogate loss; it is not a complete environment/training loop.

```python
import torch
from torch.distributions import Categorical

# Toy batch: policy logits, sampled actions, old log-probabilities, advantages.
logits = torch.tensor([[1.2, 0.1], [0.2, 0.8], [0.5, 0.4]], requires_grad=True)
actions = torch.tensor([0, 1, 0])
old_log_probs = torch.tensor([-0.35, -0.40, -0.55])
advantages = torch.tensor([1.0, 0.7, -0.3])
clip_eps = 0.2

dist = Categorical(logits=logits)
new_log_probs = dist.log_prob(actions)
ratio = torch.exp(new_log_probs - old_log_probs)
unclipped = ratio * advantages
clipped = torch.clamp(ratio, 1 - clip_eps, 1 + clip_eps) * advantages
policy_loss = -torch.minimum(unclipped, clipped).mean()
policy_loss.backward()
print(f"PPO clipped policy loss: {policy_loss.item():.4f}")
print("Gradient norm:", logits.grad.norm().item())
```

## 2. A* Search

A* prioritizes nodes by f(n) = g(n) + h(n): cost so far plus an estimated remaining cost. With an admissible heuristic, it can find an optimal path for nonnegative edge costs.

```python
import heapq

graph = {
    "A": {"B": 1, "C": 4}, "B": {"C": 2, "D": 5},
    "C": {"D": 1}, "D": {}
}
heuristic = {"A": 3, "B": 2, "C": 1, "D": 0}

def astar(graph, start, goal, h):
    frontier = [(h[start], 0, start, [start])]
    best_cost = {start: 0}
    while frontier:
        _, cost, node, path = heapq.heappop(frontier)
        if node == goal:
            return path, cost
        if cost != best_cost[node]:
            continue
        for neighbor, edge_cost in graph[node].items():
            new_cost = cost + edge_cost
            if new_cost < best_cost.get(neighbor, float("inf")):
                best_cost[neighbor] = new_cost
                heapq.heappush(frontier, (new_cost + h[neighbor],
                                          new_cost, neighbor, path + [neighbor]))
    return None, float("inf")

print(astar(graph, "A", "D", heuristic))
```

## 3. Monte Carlo Tree Search (MCTS)

MCTS repeatedly selects a promising child, expands it, simulates a rollout, and backpropagates the reward. This small single-player tree illustrates UCB selection and reward propagation; production game agents need state transitions and terminal evaluation.

```python
import math
import random

class Node:
    def __init__(self, reward=None):
        self.reward, self.visits, self.children = reward, 0, []

root = Node()
root.children = [Node(0.2), Node(0.8), Node(0.5)]
for _ in range(500):
    parent = root
    # UCB1 selection among the fixed candidate actions
    log_n = math.log(parent.visits + 1)
    child = max(parent.children, key=lambda c:
                (c.reward if c.reward is not None else 0)
                + math.sqrt(2 * log_n / (c.visits + 1)))
    # Bernoulli rollout using the candidate's reward as its win probability
    reward = 1.0 if random.random() < child.reward else 0.0
    parent.visits += 1
    child.visits += 1
print([(round(c.reward, 2), c.visits) for c in root.children])
```

## 5. Differentiable Neural Computer (DNC)

A full DNC includes differentiable external memory, content-based addressing, allocation, and read/write heads. This minimal differentiable memory sketch demonstrates soft addressing and a differentiable read; it is not a complete DNC implementation.

```python
import torch
from torch import nn
import torch.nn.functional as F

class DifferentiableMemory(nn.Module):
    def __init__(self, slots=8, width=16):
        super().__init__()
        self.memory = nn.Parameter(torch.randn(slots, width) * 0.1)
        self.key = nn.Parameter(torch.randn(width) * 0.1)
    def forward(self):
        weights = F.softmax(self.memory @ self.key, dim=0)  # soft read weights
        return weights @ self.memory, weights

memory = DifferentiableMemory()
read_vector, read_weights = memory()
loss = read_vector.square().mean()
loss.backward()
print("Read vector:", read_vector.shape, "weights sum:", read_weights.sum().item())
print("Memory gradient:", memory.memory.grad.norm().item())
```

## 6. Alpha-Beta Pruning

Alpha-beta pruning accelerates minimax by skipping branches that cannot affect the final decision. It returns the same choice as minimax when the game tree and move ordering are unchanged.

```python
def alphabeta(tree, node, depth, alpha, beta, maximizing):
    if depth == 0 or isinstance(tree[node], (int, float)):
        return tree[node]
    if maximizing:
        value = float("-inf")
        for child in tree[node]:
            value = max(value, alphabeta(tree, child, depth-1, alpha, beta, False))
            alpha = max(alpha, value)
            if alpha >= beta: break
        return value
    value = float("inf")
    for child in tree[node]:
        value = min(value, alphabeta(tree, child, depth-1, alpha, beta, True))
        beta = min(beta, value)
        if beta <= alpha: break
    return value

tree = {"root": ["left", "right"], "left": ["a", "b"],
        "right": ["c", "d"], "a": 3, "b": 5, "c": 2, "d": 9}
print("Best minimax value:", alphabeta(tree, "root", 2, float("-inf"),
                                        float("inf"), True))
```

## 8. Generative Adversarial Network (GAN)

A GAN trains a generator to create samples and a discriminator to distinguish generated samples from real data. This toy example trains both networks on a one-dimensional Gaussian; it is a teaching demo, not a stable image GAN recipe.

```python
import torch
from torch import nn

torch.manual_seed(3)
G = nn.Sequential(nn.Linear(4, 16), nn.ReLU(), nn.Linear(16, 1))
D = nn.Sequential(nn.Linear(1, 16), nn.LeakyReLU(0.2), nn.Linear(16, 1), nn.Sigmoid())
opt_g = torch.optim.Adam(G.parameters(), lr=1e-3)
opt_d = torch.optim.Adam(D.parameters(), lr=1e-3)
bce = nn.BCELoss()

for step in range(200):
    real = torch.randn(32, 1) * 0.7 + 2.0
    noise = torch.randn(32, 4)
    fake = G(noise).detach()
    d_loss = bce(D(real), torch.ones(32, 1)) + bce(D(fake), torch.zeros(32, 1))
    opt_d.zero_grad(); d_loss.backward(); opt_d.step()

    generated = G(torch.randn(32, 4))
    g_loss = bce(D(generated), torch.ones(32, 1))
    opt_g.zero_grad(); g_loss.backward(); opt_g.step()

print("Generated sample mean:", G(torch.randn(1000, 4)).mean().item())
```

## 10. Behavior Trees

Behavior trees organize decisions into composable nodes. Sequence runs children until one fails; selector runs children until one succeeds. They are common in robotics, game AI, and bounded agent workflows.

```python
SUCCESS, FAILURE, RUNNING = "SUCCESS", "FAILURE", "RUNNING"

class Sequence:
    def __init__(self, *children): self.children = children
    def tick(self, context):
        for child in self.children:
            result = child(context)
            if result != SUCCESS: return result
        return SUCCESS

def has_target(ctx): return SUCCESS if ctx.get("target_visible") else FAILURE
def within_range(ctx): return SUCCESS if ctx.get("distance", 999) < 5 else FAILURE
def engage(ctx):
    ctx["action"] = "engage_target"
    return SUCCESS

tree = Sequence(has_target, within_range, engage)
world = {"target_visible": True, "distance": 3.2}
print("Tree result:", tree.tick(world), "world:", world)
```

## 11. Soft Actor-Critic (SAC)

SAC is an off-policy actor-critic method for continuous control. It maximizes expected reward plus entropy, encouraging exploration. This example computes a simple entropy-regularized actor objective from a toy Gaussian policy; a full SAC agent also needs twin Q-networks, target networks, and replay buffer.

```python
import torch
from torch.distributions import Normal

mean = torch.tensor([0.2, -0.1], requires_grad=True)
log_std = torch.tensor([-0.5, -0.3], requires_grad=True)
distribution = Normal(mean, log_std.exp())
action = distribution.rsample()
log_prob = distribution.log_prob(action).sum()
estimated_q = -(action - 0.7).square().sum()  # toy differentiable critic
alpha = 0.2
actor_loss = (alpha * log_prob - estimated_q)
actor_loss.backward()
print("SAC-style actor loss:", actor_loss.item())
print("Policy gradient norm:", torch.stack([mean.grad.norm(), log_std.grad.norm()]).norm().item())
```

## 12. Attention Mechanisms (Self-Attention & Cross-Attention)

Self-attention lets tokens attend to other tokens in the same sequence. Cross-attention uses queries from one sequence and keys/values from another, as in encoder-decoder models and multimodal fusion.

```python
import torch
from torch import nn

torch.manual_seed(11)
self_attention = nn.MultiheadAttention(embed_dim=32, num_heads=4, batch_first=True)
cross_attention = nn.MultiheadAttention(embed_dim=32, num_heads=4, batch_first=True)
tokens = torch.randn(2, 6, 32)
context = torch.randn(2, 10, 32)

self_out, self_weights = self_attention(tokens, tokens, tokens, need_weights=True)
cross_out, cross_weights = cross_attention(tokens, context, context, need_weights=True)
print("Self-attention:", self_out.shape, "weights:", self_weights.shape)
print("Cross-attention:", cross_out.shape, "weights:", cross_weights.shape)
```

## Comparison: when to reach for each family

| Algorithm | Typical problem | Key idea | Main operational consideration |
|---|---|---|---|
| PPO | Sequential decisions | Clipped policy-gradient update | On-policy data collection can be expensive |
| A* | Route/path planning | Cost-so-far + heuristic | Heuristic quality and state-space size |
| MCTS | Large decision/game trees | Repeated simulation and UCB selection | Compute budget and rollout quality |
| LSTM | Ordered sequences | Gated recurrent state | Sequential computation and long-range memory |
| DNC | Algorithmic reasoning / structured memory | Differentiable external memory | More complex training and memory control |
| Alpha-beta | Adversarial games | Minimax with branch pruning | Move ordering and evaluation quality |
| YOLO | Object detection | One-stage box/class prediction | Model size, latency, confidence thresholds |
| GAN | Synthetic data generation | Generator vs discriminator | Training instability and evaluation |
| VGG / ResNet | Image recognition | Convolutions; residual skips | Compute, memory, transfer-learning choice |
| Behavior trees | Bounded agent/robot control | Composable condition/action nodes | Explicit failure and timeout handling |

## 4. Variance

## 6. Probability

Likelihood of an event. Estimate it theoretically from a model or empirically from repeated trials.

**Typical use:** Risk estimates, simulations, and event likelihood.


```python
def stat_06_probability(event_mask):
    event_mask = np.asarray(event_mask, dtype=bool)
    if event_mask.size == 0:
        raise ValueError('at least one trial is required')
    return float(np.mean(event_mask))

rng = np.random.default_rng(42)
dice_rolls = rng.integers(1, 7, size=20000)
win_mask = np.isin(dice_rolls, [5, 6])
empirical_probability = stat_06_probability(win_mask)
theoretical_probability = 2 / 6
print(f'Theoretical P(5 or 6): {theoretical_probability:.3f}')
print(f'Empirical P(5 or 6):  {empirical_probability:.3f}')
```

```python
trial_counts = np.cumsum(win_mask)
trial_rates = trial_counts / np.arange(1, len(win_mask)+1)
plt.figure(figsize=(8, 4)); plt.plot(trial_rates, label='Running empirical probability')
plt.axhline(theoretical_probability, color='crimson', linestyle='--', label='Theoretical probability')
plt.title('6. Probability — convergence in a dice simulation'); plt.xlabel('Number of trials'); plt.ylabel('Estimated probability'); plt.legend(); plt.tight_layout(); plt.show()
```

## 7. Bayes’ Theorem

Updates a prior probability using evidence: P(A|B) = P(B|A)P(A) / P(B).

**Typical use:** Diagnostic tests, spam filtering, and evidence updates.


```python
def stat_07_bayes(prior, sensitivity, specificity):
    if not all(0 <= x <= 1 for x in (prior, sensitivity, specificity)):
        raise ValueError('all probabilities must be between 0 and 1')
    false_positive = 1 - specificity
    p_positive = sensitivity*prior + false_positive*(1-prior)
    return 0.0 if p_positive == 0 else sensitivity*prior/p_positive

prior, sensitivity, specificity = 0.01, 0.95, 0.90
posterior = stat_07_bayes(prior, sensitivity, specificity)
print(f'P(condition | positive test): {posterior:.2%}')
```

```python
labels = ['Prior probability', 'Posterior after positive']
values = [prior, posterior]
plt.figure(figsize=(7, 4)); plt.bar(labels, values)
plt.ylim(0, 1); plt.ylabel('Probability'); plt.title('7. Bayes’ theorem — updating belief with evidence')
for i, v in enumerate(values): plt.text(i, v + .025, f'{v:.1%}', ha='center')
plt.tight_layout(); plt.show()
```

## 8. Correlation

Measures direction and strength of linear association. Correlation alone does not establish causation.

**Typical use:** Feature analysis and exploratory data analysis.


```python
def stat_08_correlation(x, y):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    if x.size != y.size or x.size < 2:
        raise ValueError('x and y must have equal length >= 2')
    if np.std(x) == 0 or np.std(y) == 0:
        raise ValueError('correlation is undefined for a constant input')
    return float(np.corrcoef(x, y)[0, 1])

hours = np.array([1, 2, 3, 4, 5, 6], dtype=float)
scores = np.array([52, 58, 65, 70, 78, 85], dtype=float)
correlation = stat_08_correlation(hours, scores)
print(f'Pearson correlation: {correlation:.3f}')
```

```python
plt.figure(figsize=(7, 4)); plt.scatter(hours, scores, s=65)
fit_line = np.polyfit(hours, scores, 1); plt.plot(hours, np.polyval(fit_line, hours), '--', label=f'r = {correlation:.2f}')
plt.title('8. Correlation — study time and exam score'); plt.xlabel('Study hours'); plt.ylabel('Exam score'); plt.legend(); plt.tight_layout(); plt.show()
```

## 9. Regression

Fits a model to predict a continuous target. This example solves ordinary least squares with NumPy.

**Typical use:** House prices, demand, and numeric forecasting.


```python
def stat_09_linear_regression(x, y):
    x, y = np.asarray(x, float).ravel(), np.asarray(y, float).ravel()
    if x.size != y.size or x.size < 2:
        raise ValueError('x and y must have equal length >= 2')
    design = np.column_stack([np.ones(x.size), x])
    coefficients, *_ = np.linalg.lstsq(design, y, rcond=None)
    return float(coefficients[0]), float(coefficients[1])

reg_x = np.array([1, 2, 3, 4, 5, 6], dtype=float)
reg_y = np.array([52, 58, 65, 70, 78, 85], dtype=float)
reg_intercept, reg_slope = stat_09_linear_regression(reg_x, reg_y)
print(f'Intercept: {reg_intercept:.2f}; slope: {reg_slope:.2f}')
print(f'Prediction at 7 hours: {reg_intercept + reg_slope*7:.2f}')
```

```python
reg_predictions = reg_intercept + reg_slope*reg_x
plt.figure(figsize=(7, 4)); plt.scatter(reg_x, reg_y, label='Observed')
plt.plot(reg_x, reg_predictions, color='crimson', label='OLS regression line')
plt.title('9. Linear regression — fitted relationship'); plt.xlabel('Study hours'); plt.ylabel('Exam score'); plt.legend(); plt.tight_layout(); plt.show()
```

## 10. Normal Distribution

A continuous bell-shaped distribution defined by its mean and standard deviation.

**Typical use:** Natural measurements and statistical modeling when assumptions are reasonable.


```python
def stat_10_normal_pdf(x, mean=0.0, std=1.0):
    if std <= 0:
        raise ValueError('std must be positive')
    x = np.asarray(x, dtype=float)
    z = (x-mean)/std
    return np.exp(-0.5*z*z)/(std*np.sqrt(2*np.pi))

normal_sample = np.random.default_rng(10).normal(100, 15, 1000)
print(f'Sample mean: {normal_sample.mean():.2f}; sample SD: {normal_sample.std(ddof=1):.2f}')
```

```python
normal_x = np.linspace(40, 160, 400)
plt.figure(figsize=(8, 4)); plt.hist(normal_sample, bins=30, density=True, alpha=.45, label='Simulated sample')
plt.plot(normal_x, stat_10_normal_pdf(normal_x, 100, 15), linewidth=2, label='Normal PDF')
plt.title('10. Normal distribution'); plt.xlabel('Value'); plt.ylabel('Density'); plt.legend(); plt.tight_layout(); plt.show()
```

## 11. Hypothesis Testing

A formal procedure for evaluating a null hypothesis against sample evidence. The decision depends on a test statistic, p-value, and significance level.

**Typical use:** Validating claims and controlled experiments.


```python
def stat_11_one_sample_ttest(sample, null_mean=0.0):
    from scipy import stats
    sample = np.asarray(sample, dtype=float)
    if sample.size < 2:
        raise ValueError('sample must contain at least two values')
    return stats.ttest_1samp(sample, popmean=null_mean)

test_sample = np.array([104, 101, 99, 107, 103, 105, 102, 106], dtype=float)
test_result = stat_11_one_sample_ttest(test_sample, 100)
print(f't = {test_result.statistic:.3f}; p = {test_result.pvalue:.4f}')
print('Decision at alpha=.05:', 'Reject H0' if test_result.pvalue < .05 else 'Fail to reject H0')
```

```python
plt.figure(figsize=(8, 4)); plt.hist(test_sample, bins=6, alpha=.6, edgecolor='black')
plt.axvline(100, color='crimson', linestyle='--', label='Null mean = 100')
plt.axvline(test_sample.mean(), color='darkgreen', linestyle='-', label=f'Sample mean = {test_sample.mean():.1f}')
plt.title('11. Hypothesis test — sample evidence vs null value'); plt.xlabel('Observed value'); plt.ylabel('Frequency'); plt.legend(); plt.tight_layout(); plt.show()
```

## 12. P-value

Assuming the null hypothesis is true, the p-value is the probability of a result at least as extreme as the observed result.

**Typical use:** Assessing statistical evidence; it is not the probability that the null hypothesis is true.


```python
def stat_12_pvalue(sample, null_mean=0.0):
    from scipy import stats
    sample = np.asarray(sample, dtype=float)
    if sample.size < 2:
        raise ValueError('sample must contain at least two values')
    return float(stats.ttest_1samp(sample, popmean=null_mean).pvalue)

pvalue_sample = np.array([104, 101, 99, 107, 103, 105, 102, 106], dtype=float)
pvalue = stat_12_pvalue(pvalue_sample, 100)
print(f'p-value: {pvalue:.5f}')
```

```python
plt.figure(figsize=(7, 3.5)); plt.bar(['Observed p-value', 'Significance level'], [pvalue, .05])
plt.axhline(.05, color='crimson', linestyle='--'); plt.ylim(0, max(.06, pvalue*1.2))
plt.title('12. P-value compared with alpha = 0.05'); plt.ylabel('Probability threshold'); plt.tight_layout(); plt.show()
```

## 13. Confidence Interval

An interval-estimation procedure. A 95% confidence method captures the true parameter in 95% of repeated samples under its assumptions.

**Typical use:** Communicating uncertainty around an estimated mean.


```python
def stat_13_mean_ci(sample, confidence=.95):
    from scipy import stats
    sample = np.asarray(sample, dtype=float)
    if sample.size < 2 or not 0 < confidence < 1:
        raise ValueError('need >=2 observations and confidence between 0 and 1')
    return stats.t.interval(confidence, df=sample.size-1, loc=sample.mean(), scale=stats.sem(sample))

ci_sample = np.array([12, 14, 13, 15, 11, 16, 14, 13], dtype=float)
ci_low, ci_high = stat_13_mean_ci(ci_sample)
print(f'Mean: {ci_sample.mean():.2f}; 95% CI: [{ci_low:.2f}, {ci_high:.2f}]')
```

```python
plt.figure(figsize=(8, 2.8)); plt.errorbar(ci_sample.mean(), 0, xerr=[[ci_sample.mean()-ci_low], [ci_high-ci_sample.mean()]], fmt='o', capsize=8, markersize=8)
plt.yticks([]); plt.xlabel('Estimated population mean'); plt.title('13. 95% confidence interval'); plt.tight_layout(); plt.show()
```

## 14. T-test

Compares means. Welch’s independent-samples t-test does not assume equal population variances.

**Typical use:** Comparing two independent groups.


```python
def stat_14_welch_ttest(group_a, group_b):
    from scipy import stats
    a, b = np.asarray(group_a, float), np.asarray(group_b, float)
    if a.size < 2 or b.size < 2:
        raise ValueError('each group needs at least two observations')
    return stats.ttest_ind(a, b, equal_var=False)

t_group_a = np.array([78, 82, 85, 88, 90, 84], dtype=float)
t_group_b = np.array([72, 75, 79, 80, 77, 74], dtype=float)
t_result = stat_14_welch_ttest(t_group_a, t_group_b)
print(f'Welch t = {t_result.statistic:.3f}; p = {t_result.pvalue:.4f}')
```

```python
plt.figure(figsize=(7, 4)); plt.boxplot([t_group_a, t_group_b], tick_labels=['Group A', 'Group B'])
plt.title('14. Welch t-test — compare two group distributions'); plt.ylabel('Measured value'); plt.tight_layout(); plt.show()
```

## 15. ANOVA

One-way analysis of variance compares means across three or more groups. A significant result indicates at least one mean differs.

**Typical use:** Comparing treatment groups or multiple model variants.


```python
def stat_15_one_way_anova(*groups):
    from scipy import stats
    arrays = [np.asarray(g, dtype=float) for g in groups]
    if len(arrays) < 2 or any(a.size < 2 for a in arrays):
        raise ValueError('provide at least two groups with two values each')
    return stats.f_oneway(*arrays)

anova_control = np.array([10, 11, 9, 10, 12], float)
anova_a = np.array([13, 14, 12, 15, 13], float)
anova_b = np.array([16, 15, 17, 14, 16], float)
anova_result = stat_15_one_way_anova(anova_control, anova_a, anova_b)
print(f'F = {anova_result.statistic:.3f}; p = {anova_result.pvalue:.5f}')
```

```python
plt.figure(figsize=(7, 4)); plt.boxplot([anova_control, anova_a, anova_b], tick_labels=['Control', 'Treatment A', 'Treatment B'])
plt.title('15. ANOVA — compare multiple group means'); plt.ylabel('Outcome'); plt.tight_layout(); plt.show()
```

## 16. Chi-Square Test

Tests whether two categorical variables are independent using observed and expected contingency-table counts.

**Typical use:** Survey analysis, click-through experiments, and categorical associations.


```python
def stat_16_chi_square(observed):
    from scipy import stats
    observed = np.asarray(observed, dtype=int)
    if observed.ndim != 2 or np.any(observed < 0) or np.any(observed.sum(axis=0) == 0) or np.any(observed.sum(axis=1) == 0):
        raise ValueError('observed must be a nonnegative 2D contingency table with nonempty rows/columns')
    return stats.chi2_contingency(observed)

chi_table = np.array([[30, 50], [70, 50]])
chi_result = stat_16_chi_square(chi_table)
print(f'Chi-square = {chi_result.statistic:.3f}; dof = {chi_result.dof}; p = {chi_result.pvalue:.4f}')
print('Expected counts:\n', chi_result.expected_freq)
```

```python
plt.figure(figsize=(6, 4)); plt.imshow(chi_table, cmap='Blues')
for (i, j), value in np.ndenumerate(chi_table): plt.text(j, i, str(value), ha='center', va='center')
plt.xticks([0, 1], ['Control', 'New design']); plt.yticks([0, 1], ['Clicked', 'Not clicked'])
plt.title('16. Chi-square — observed contingency counts'); plt.colorbar(label='Count'); plt.tight_layout(); plt.show()
```

## 18. K-Means Clustering

Partitions observations into k clusters by iteratively assigning points to nearest centroids and updating centroids.

**Typical use:** Customer segmentation and grouping compact, roughly spherical clusters.


```python
def stat_18_kmeans(X, k=3, iterations=50, seed=42):
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or not 1 <= k <= len(X):
        raise ValueError('X must be 2D and k between 1 and number of rows')
    rng = np.random.default_rng(seed)
    centers = X[rng.choice(len(X), size=k, replace=False)].copy()
    for _ in range(iterations):
        labels = np.argmin(((X[:, None, :] - centers[None, :, :])**2).sum(axis=2), axis=1)
        updated = np.array([X[labels == j].mean(axis=0) if np.any(labels == j) else centers[j] for j in range(k)])
        if np.allclose(updated, centers): break
        centers = updated
    inertia = float(((X - centers[labels])**2).sum())
    return labels, centers, inertia

cluster_rng = np.random.default_rng(18)
cluster_X = np.vstack([cluster_rng.normal(loc=c, scale=.55, size=(70, 2)) for c in [(-2, -1), (2, 0), (0, 3)]])
cluster_labels, cluster_centers, cluster_inertia = stat_18_kmeans(cluster_X, k=3)
print(f'Inertia: {cluster_inertia:.2f}; centers:\n{cluster_centers}')
```

```python
plt.figure(figsize=(7, 4)); plt.scatter(cluster_X[:, 0], cluster_X[:, 1], c=cluster_labels, s=22)
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], marker='X', s=180, edgecolor='black', label='Centroids')
plt.title('18. K-Means — clusters and centroids'); plt.xlabel('Feature 1'); plt.ylabel('Feature 2'); plt.legend(); plt.tight_layout(); plt.show()
```

## 19. Principal Component Analysis (PCA)

Reduces dimensionality by projecting centered data onto orthogonal directions of maximum variance. This example uses NumPy SVD.

**Typical use:** Visualization, compression, and reducing correlated features.


```python
def stat_19_pca(X, n_components=2):
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or not 1 <= n_components <= min(X.shape):
        raise ValueError('invalid matrix shape or component count')
    mean = X.mean(axis=0)
    centered = X - mean
    _, singular_values, components = np.linalg.svd(centered, full_matrices=False)
    scores = centered @ components[:n_components].T
    variance = singular_values**2 / (len(X)-1)
    explained = variance / variance.sum()
    return scores, components[:n_components], explained[:n_components]

pca_rng = np.random.default_rng(19)
pca_base = pca_rng.normal(size=(180, 3))
pca_data = np.column_stack([pca_base, 2*pca_base[:, 0] - .5*pca_base[:, 1] + .1*pca_rng.normal(size=180)])
pca_scores, pca_components, pca_explained = stat_19_pca(pca_data, 2)
print('Original shape:', pca_data.shape, '| projected shape:', pca_scores.shape)
print('Explained variance ratio:', pca_explained)
```

```python
plt.figure(figsize=(7, 4)); plt.scatter(pca_scores[:, 0], pca_scores[:, 1], alpha=.7, s=24)
plt.title(f'19. PCA — first two components retain {pca_explained.sum():.1%} variance')
plt.xlabel('Principal component 1'); plt.ylabel('Principal component 2'); plt.tight_layout(); plt.show()
```

## 20. Time Series Forecasting

Uses ordered historical observations to predict future values. This baseline fits a linear trend to past observations only.

**Typical use:** Demand, sales, sensor readings, and operational metrics.


```python
def stat_20_time_series_forecast(values, horizon=6):
    values = np.asarray(values, dtype=float).ravel()
    if values.size < 3 or horizon < 1:
        raise ValueError('need at least 3 observations and a positive horizon')
    t = np.arange(values.size, dtype=float)
    slope, intercept = np.polyfit(t, values, 1)
    future_t = np.arange(values.size, values.size+horizon, dtype=float)
    return intercept + slope*future_t, float(slope), float(intercept)

series_rng = np.random.default_rng(20)
series_values = 100 + .8*np.arange(48) + 8*np.sin(2*np.pi*np.arange(48)/12) + series_rng.normal(0, 1.5, 48)
series_forecast, series_slope, series_intercept = stat_20_time_series_forecast(series_values, horizon=12)
print(f'Estimated trend per period: {series_slope:.2f}')
print('Next 12 forecasts:', np.round(series_forecast, 2))
```

```python
all_t = np.arange(len(series_values)); future_t = np.arange(len(series_values), len(series_values)+len(series_forecast))
plt.figure(figsize=(9, 4)); plt.plot(all_t, series_values, marker='.', label='Historical observations')
plt.plot(future_t, series_forecast, marker='o', linestyle='--', label='Linear-trend forecast')
plt.axvline(len(series_values)-1, color='gray', linestyle=':'); plt.title('20. Time series forecasting — historical data and future trend')
plt.xlabel('Time period'); plt.ylabel('Value'); plt.legend(); plt.tight_layout(); plt.show()
```
