# Architecture Compiler — Interactive POC
A notebook to inspect the YAML IR, policy checks, and backends. Run from the repo root after `pip install -e .`.

````
from pathlib import Path
import yaml, json
from archcompiler.compile import load_spec, check_constraints, compile_target
source = Path("examples/application.yaml")
spec = load_spec(source)
print(yaml.safe_dump(spec.model_dump(by_alias=True, exclude_none=True), sort_keys=False))
````

````
print("Violations:", check_constraints(spec))
print(compile_target(spec, "rego"))
````

````
spec.resources["database"].network_access = "public"
print("After deliberately violating a mandatory rule:", check_constraints(spec))
````

````
network = load_spec("examples/network.yaml")
print(json.dumps(json.loads(compile_target(network, "terraform")), indent=2))
````
